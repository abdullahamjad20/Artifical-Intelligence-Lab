```python
# ===== Question 1 (self-contained: no matplotlib needed) =====
import heapq, time

GOAL = (1, 2, 3, 4, 5, 6, 7, 8, 0)
MOVES = {'Up': -3, 'Down': 3, 'Left': -1, 'Right': 1}

def neighbors(state):
    z = state.index(0)
    r, c = divmod(z, 3)
    for name, d in MOVES.items():
        if (name == 'Up' and r == 0) or (name == 'Down' and r == 2) \
           or (name == 'Left' and c == 0) or (name == 'Right' and c == 2):
            continue
        s = list(state)
        s[z], s[z + d] = s[z + d], s[z]
        yield name, tuple(s)

def manhattan(state):
    t = 0
    for i, v in enumerate(state):
        if v:
            g = GOAL.index(v)
            t += abs(i // 3 - g // 3) + abs(i % 3 - g % 3)
    return t

def is_solvable(state):
    t = [x for x in state if x]
    inv = sum(1 for i in range(len(t)) for j in range(i + 1, len(t)) if t[i] > t[j])
    return inv % 2 == 0

def solve(start):
    if not is_solvable(start):
        return None, None
    t0 = time.perf_counter()
    tie = 0
    open_heap = [(manhattan(start), 0, tie, start)]
    parent = {start: None}
    g_cost = {start: 0}
    expanded, generated = 0, 1
    closed = set()
    while open_heap:
        f, g, _, cur = heapq.heappop(open_heap)
        if cur in closed:
            continue
        closed.add(cur)
        if cur == GOAL:
            path = []
            while cur is not None:
                path.append(cur)
                cur = parent[cur]
            path.reverse()
            return path, dict(length=len(path) - 1, expanded=expanded,
                              generated=generated, time=time.perf_counter() - t0)
        expanded += 1
        for _, nxt in neighbors(cur):
            ng = g + 1
            if nxt not in g_cost or ng < g_cost[nxt]:
                g_cost[nxt] = ng
                parent[nxt] = cur
                generated += 1
                tie += 1
                heapq.heappush(open_heap, (ng + manhattan(nxt), ng, tie, nxt))
    return None, None

start = (8, 1, 3, 4, 0, 2, 7, 6, 5)
path, st = solve(start)

if st is None:
    print("Start state is unsolvable.")
else:
    print("=== Q1: start", start)
    print(f"Solution length : {st['length']}")
    print(f"Nodes expanded  : {st['expanded']}")
    print(f"Nodes generated : {st['generated']}")
    print(f"Time            : {st['time']:.4f} s")
```
